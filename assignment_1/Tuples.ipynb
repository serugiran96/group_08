# Tuples

```python
ages = (20, 25, 22, 30, 28)
print("Tupla:", ages)
```

```python
# The first and last elements.
first_age = ages[0]
last_age = ages[-1]
print("First age:", first_age)
print("Last age:", last_age)
```

```python
# The maximum value; the minimum value and the number of elements.

max_age = max(ages)
min_age = min(ages) 
print("Maximum age:", max_age)
print("Minimum age:", min_age)
```
