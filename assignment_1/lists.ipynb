```python
grades = [15, 18, 12, 20, 16] #Print the original list
print (grades)
```

```python
grades.append(17) #Add the grade 17 using append()
print (grades)
```

```python
grades_sort = grades.copy() #Create a copy of the list 
grades_sort
```

```python
grades_sort.sort() #Sort the copy from lowest to highest using sort()
print(grades_sort)
```

```python
print("Highest grade:", max(grades_sort)) #the highest grade
print("Lowest grade:", min(grades_sort)) #the lowest grade
print("Total number of grades:", len(grades_sort)) #the total number of grades
```

```python
print(grades_sort) #Print the final list
```
