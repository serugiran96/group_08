
# Visualización interactiva: lluvias y declaratorias de emergencia en el Perú

## Pregunta de investigación

¿El Estado declara emergencia donde más llueve?

El objetivo es explorar la relación entre las precipitaciones registradas y las declaratorias de emergencia por lluvias en los departamentos del Perú, durante el período comprendido entre el 1 de enero y el 31 de mayo de 2025.

```python
import pandas as pd
import plotly.express as px

datos = pd.read_csv(
    "datos/dataset.csv",
    dtype={"ubigeo": str},
    encoding="utf-8-sig"
)

print("Dimensiones:", datos.shape)

display(datos.head())
```

```python
FUENTE = (
    "Fuente: Decretos supremos de la PCM (gob.pe), periodo enero a mayo de 2025."
)
```


## 1. Gráfico de barras: declaratorias de emergencia por departamento

El siguiente gráfico compara el número de declaratorias de emergencia por lluvias registradas en cada departamento entre enero y mayo del año 2025.

Las barras horizontales permiten comparar los departamentos e identificar aquellos con más declaratorias.

```python

# Ordenar los departamentos por número de declaratorias
orden = datos.sort_values("declaratorias")

# Crear el gráfico de barras horizontales
fig_barras = px.bar(
    orden,
    x="declaratorias",
    y="departamento",
    orientation="h",
    title=(
        "Cusco concentra el mayor número de declaratorias por lluvias: "
        "8 departamentos se igualan con 8"
    ),
    labels={
        "declaratorias": "Declaratorias de emergencia (enero a mayo de 2025)",
        "departamento": ""
    },
    template="plotly_white",
    height=650
)

# Agregar la fuente debajo del gráfico
fig_barras.add_annotation(
    text=FUENTE,
    xref="paper",
    yref="paper",
    x=0,
    y=-0.1,
    showarrow=False,
    xanchor="left",
    font_size=11
)

# Ajustar el espacio inferior para que se lea la fuente
fig_barras.update_layout(
    margin=dict(b=90)
)

# Mostrar el gráfico interactivo
fig_barras.show()
```

```python
FUENTE = (
    "Fuente: Decretos supremos de la PCM (gob.pe), periodo enero a mayo de 2025."
)
```

Nota: Considerar que los datos corresponden a la lluvia acumulada en la capital departamental, por lo que no debe asumirse como la lluvia total de todo el departamento.


## 2. Gráfico de dispersión: lluvia acumulada y declaratorias de emergencia

El gráfico explora la relación entre la lluvia acumulada en las capitales departamentales y el número de declaratorias de emergencia por lluvias, entre enero y mayo del año 2025.

Cada punto representa un departamento. Al pasar el cursor sobre él, se puede consultar el nombre del departamento, su capital y los días con lluvia fuerte.

Asimismo, se puede identificar casos que se apartan de una posible relación general entre las precipitaciones y las declaratorias.


```python

fig_dispersion = px.scatter(
    datos,
    x="lluvia_total_mm",
    y="declaratorias",
    hover_name="departamento",
    hover_data={
        "capital": True,
        "dias_lluvia_fuerte": True
    },
    title=(
        "Mayor concentración de lluvia no siempre se traduce en más declaratorias"
    ),
    labels={
        "lluvia_total_mm": "Lluvia acumulada en la capital (mm)",
        "declaratorias": "Declaratorias de emergencia"
    },
    template="plotly_white"
)

# Aumentar el tamaño de los puntos
fig_dispersion.update_traces(
    marker={"size": 11}
)

# Agregar la fuente
fig_dispersion.add_annotation(
    text=FUENTE,
    xref="paper",
    yref="paper",
    x=0,
    y=-0.15,
    showarrow=False,
    xanchor="left",
    font_size=11
)

# Reservar espacio para la fuente
fig_dispersion.update_layout(
    margin=dict(b=90)
)

# Mostrar el gráfico
fig_dispersion.show()
```

```python
FUENTE = (
    "Fuente: Decretos supremos de la PCM (gob.pe), periodo enero a mayo de 2025."
)
```


## 3. Histograma: distribución de los días con lluvia fuerte

El presente gráfico muestra cómo se distribuyen las capitales departamentales según la cantidad de días con precipitaciones de 20 mm o más, entre enero y mayo del año 2025.

De esta manera, el histograma permite identificar los rangos de días más frecuentes y observar si algunas capitales registraron muchos más días de lluvia fuerte que el resto.


```python

fig_hist = px.histogram(
    datos,
    x="dias_lluvia_fuerte",
    nbins=10,
    title=(
        "La mayoría de las capitales registró pocos días "
        "de lluvia fuerte"
    ),
    labels={
        "dias_lluvia_fuerte": "Días con 20 mm o más",
        "count": "Número de capitales"
    },
    template="plotly_white"
)

# Etiquetar el eje vertical
fig_hist.update_layout(
    yaxis_title="Número de capitales",
    margin=dict(b=90)
)

# Agregar la fuente
fig_hist.add_annotation(
    text=FUENTE,
    xref="paper",
    yref="paper",
    x=0,
    y=-0.15,
    showarrow=False,
    xanchor="left",
    font_size=11
)

# Mostrar el histograma
fig_hist.show()
```

```python
FUENTE = (
    "Fuente: Decretos supremos de la PCM (gob.pe), periodo enero a mayo de 2025."
)
```


## 4. Verificación de los valores extremos

A continuación, se revisan los departamentos con los valores más altos y más bajos de las principales variables de análisis. Esta comprobación permite contrastar los resultados de la tabla con lo que muestran los gráficos y detectar posibles errores de interpretación.


```python
print("DEPARTAMENTOS CON MÁS DECLARATORIAS")
display(
    datos.nlargest(3, "declaratorias")[
        ["departamento", "declaratorias"]
    ]
)

print("DEPARTAMENTOS CON MÁS LLUVIA")
display(
    datos.nlargest(3, "lluvia_total_mm")[
        ["departamento", "lluvia_total_mm"]
    ]
)

print("DEPARTAMENTOS CON MENOS LLUVIA")
display(
    datos.nsmallest(3, "lluvia_total_mm")[
        ["departamento", "lluvia_total_mm"]
    ]
)

print("DEPARTAMENTOS CON MÁS DÍAS DE LLUVIA FUERTE")
display(
    datos.nlargest(3, "dias_lluvia_fuerte")[
        ["departamento", "dias_lluvia_fuerte"]
    ]
)
```


### Comprobación e interpretación

Los valores extremos de la tabla coinciden con los resultados observados en los gráficos. Cusco registra el mayor número de declaratorias de emergencia, con 9, seguido de ocho departamentos que registran 8 cada uno.

En relación a la lluvia acumulada en las capitales de los departamentos, Loreto y Madre de Dios concentran los valores más altos. Del mismo modo, Madre de Dios también presenta el mayor número de días con lluvia fuerte.

Estos resultados sugieren que una mayor cantidad de lluvia acumulada no siempre significa un mayor número de declaratorias de emergencia.
