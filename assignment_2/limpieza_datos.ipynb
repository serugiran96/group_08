# Parte 2 - Limpieza de datos

Revisamos y limpiamos los datos obtenidos mediante scraping.

```python
import pandas as pd

df = pd.read_csv("datos/decretos_lluvias.csv")
df.head()
```

## Paso 1: Revisar los datos

```python
df.info()
```

```python
print("Duplicados:", df.duplicated().sum())

print("\nValores de tipo:")
print(df["tipo"].value_counts())

print("\nValores de motivo:")
print(df["motivo"].value_counts())
```

## Paso 2: Limpiar los datos

```python
# Convertir los meses a números
meses = {
    "enero": "01",
    "febrero": "02",
    "marzo": "03",
    "abril": "04",
    "mayo": "05",
    "junio": "06",
    "julio": "07",
    "agosto": "08",
    "septiembre": "09",
    "octubre": "10",
    "noviembre": "11",
    "diciembre": "12"
}

for mes, numero in meses.items():
    df["fecha"] = df["fecha"].str.replace(f" de {mes} de ", f"/{numero}/")

# Convertir la columna a fecha
df["fecha"] = pd.to_datetime(df["fecha"], format="%d/%m/%Y")

# Eliminar duplicados y valores nulos
df = df.drop_duplicates()
df = df.dropna()

df.head()
```

## Paso 3: Verificar los datos limpios

```python
df.info()

print("\nDuplicados:", df.duplicated().sum())
print("Valores nulos:")
print(df.isnull().sum())
```

## Paso 4: Comparar antes y después

```python
antes = pd.read_csv("datos/decretos_lluvias.csv")

comparacion = pd.DataFrame({
    "fecha_antes": antes["fecha"],
    "fecha_despues": df["fecha"]
})

comparacion.head()
```

## Paso 5: Guardar los datos limpios

```python
df.to_csv("datos/decretos_lluvias_limpio.csv", index=False)
```
